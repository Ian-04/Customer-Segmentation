```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("D:/Projects/Kaggle/credit_scroring/CC GENERAL.csv")
data
```

EDA

Spending behaviour

```python
# Purchases
plt.figure(figsize=(16,6))
sns.histplot(
    data = data,
    x = data['PURCHASES'],
    color = 'orange'
    #bins = 10
    #kde=True
)

plt.xlabel('Purchases')
plt.ylabel('Frequency'),
plt.title('Customer Purchase Distribution')
```

From the above distribution, it suggest most customers do approximately 5K purchases. There also those that do high purchases around 48k but these users are significatly few.

```python
#One off Purchases
plt.figure(figsize=(16,5))
sns.boxenplot(
    data = data,
    x = data['ONEOFF_PURCHASES']
)
plt.xlabel('One Off Purchases')
plt.ylabel('Frequency')
plt.title('Distribution of One off Purchases')
           
```

Most customers are doing few one off purchases

```python
#Number of purchase transactios
plt.figure(figsize=(16,6))
sns.kdeplot(
    data = data,
    x = data['PURCHASES_TRX'],
    fill=True,
    color = 'darkgreen'
)

plt.xlabel('Purchase Transactions')
plt.ylabel('Frequency')
plt.title('Distribution of Purchase Transactions')
```

Usage Frequency

```python
# Purchase Frequency
plt.figure(figsize=(16,6))
sns.histplot(
    data = data,
    x = data['PURCHASES_FREQUENCY'],
)

plt.xlabel('Purchase Frequency')
plt.ylabel('Frequency')
plt.title('Distribution of Purchase Frequency')
```

```python
# Balance frequency
plt.figure(figsize=(16,6))
sns.boxplot(
    data = data,
    x = data['BALANCE_FREQUENCY']
)

plt.xlabel('Balance Frequency')
plt.ylabel('Density')
plt.title('Distribution of Balance Frequency')
```

Most of the customers frequently updates their balance

```python
#Number of purchase transactios
plt.figure(figsize=(16,6))
sns.kdeplot(
    data = data,
    x = data['PAYMENTS'],
    fill=True,
    color = 'skyblue'
)

plt.xlabel('Payments')
plt.ylabel('Frequency')
plt.title('Distribution of Payments')
```
